```python
%%writefile src/train_uce.py
import os
import torch
import torch.nn as nn
from diffusers import StableDiffusionPipeline

MODEL_ID = "CompVis/stable-diffusion-v1-4"
REVISION = "133a221"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float32  # Use float32 for stable matrix operations

# -------------------------------------------------------------
# 1. DEFINE CONCEPTS & TARGETS
# -------------------------------------------------------------
# Styles to erase -> map towards generic art
STYLE_EDITS = [
    # Wassily Kandinsky
    ("in the style of Wassily Kandinsky", "a painting"),
    ("Wassily Kandinsky", "art"),
    ("a painting by Wassily Kandinsky", "a painting"),
    # Gustav Klimt
    ("in the style of Gustav Klimt", "a painting"),
    ("Gustav Klimt", "art"),
    ("a painting by Gustav Klimt", "a painting"),
    # Edvard Munch
    ("in the style of Edvard Munch", "a painting"),
    ("Edvard Munch", "art"),
    ("a painting by Edvard Munch", "a painting"),
    # Frida Kahlo
    ("in the style of Frida Kahlo", "a painting"),
    ("Frida Kahlo", "art"),
    ("a painting by Frida Kahlo", "a painting"),
    # Amedeo Modigliani
    ("in the style of Amedeo Modigliani", "a painting"),
    ("Amedeo Modigliani", "art"),
    ("a painting by Amedeo Modigliani", "a painting"),
]

# Roles to debias -> map towards 50/50 male & female blend
ROLE_DEBIAS = [
    ("CEO", "a female CEO", "a male CEO"),
    ("a candid photograph of a CEO such that face is visible",
     "a candid photograph of a female CEO such that face is visible",
     "a candid photograph of a male CEO such that face is visible"),
    ("nurse", "a female nurse", "a male nurse"),
    ("a candid photograph of a nurse such that face is visible",
     "a candid photograph of a female nurse such that face is visible",
     "a candid photograph of a male nurse such that face is visible"),
    ("construction worker", "a female construction worker", "a male construction worker"),
    ("a candid photograph of a construction worker such that face is visible",
     "a candid photograph of a female construction worker such that face is visible",
     "a candid photograph of a male construction worker such that face is visible"),
    ("secretary", "a female secretary", "a male secretary"),
    ("a candid photograph of a secretary such that face is visible",
     "a candid photograph of a female secretary such that face is visible",
     "a candid photograph of a male secretary such that face is visible"),
    ("chef", "a female chef", "a male chef"),
    ("a candid photograph of a chef such that face is visible",
     "a candid photograph of a female chef such that face is visible",
     "a candid photograph of a male chef such that face is visible"),
]

# Control / Preservation Prompts (Protects Style Preserver & Bias Preserver)
PRESERVE_PROMPTS = [
    "a painting of a cat, impressionist style",
    "a watercolor of a mountain lake",
    "a charcoal sketch of an old bridge",
    "a pop-art poster of a city skyline",
    "an ink illustration of a dragon",
    "a photo of a librarian at a desk",
    "a photo of a pilot in a cockpit",
    "a photo of a teacher in a classroom",
    "a photo of a farmer in a field",
    "a photo of a violinist performing",
    "a photo of a doctor",
    "a portrait of a smiling person",
    "oil painting of flowers in a vase",
    "a landscape with trees and river",
    "a realistic photograph of an athlete",
    "an architectural sketch of a building"
]

def get_text_embedding(pipe, text: str) -> torch.Tensor:
    """Extract mean non-padding text embedding vector (shape: 768)."""
    tokens = pipe.tokenizer(
        text,
        padding="max_length",
        max_length=pipe.tokenizer.model_max_length,
        truncation=True,
        return_tensors="pt"
    ).input_ids.to(DEVICE)
    
    with torch.no_grad():
        emb = pipe.text_encoder(tokens)[0][0] # (77, 768)
        
    # Average across non-padding tokens
    non_pad_len = min(len(pipe.tokenizer.encode(text)), 77)
    return emb[1:non_pad_len-1].mean(dim=0) if non_pad_len > 2 else emb[1]

def main():
    print("Loading base Stable Diffusion pipeline...")
    pipe = StableDiffusionPipeline.from_pretrained(
        MODEL_ID,
        revision=REVISION,
        torch_dtype=DTYPE
    ).to(DEVICE)

    # -------------------------------------------------------------
    # 2. COLLECT KEYS (K) AND TARGET VALUE REPRESENTATIONS (V)
    # -------------------------------------------------------------
    print("Computing concept conditioning vectors...")
    keys = []
    values_raw = []  # We store the target text vectors to be projected by W_orig

    # Style Erasure
    for src, dst in STYLE_EDITS:
        k_src = get_text_embedding(pipe, src)
        k_dst = get_text_embedding(pipe, dst)
        keys.append(k_src)
        values_raw.append(('single', k_dst))

    # Role Debiasing
    for src, female, male in ROLE_DEBIAS:
        k_src = get_text_embedding(pipe, src)
        k_fem = get_text_embedding(pipe, female)
        k_male = get_text_embedding(pipe, male)
        keys.append(k_src)
        values_raw.append(('blend', k_fem, k_male))

    # Concept Preservation
    for pres in PRESERVE_PROMPTS:
        k_pres = get_text_embedding(pipe, pres)
        keys.append(k_pres)
        values_raw.append(('preserve', k_pres))

    # K matrix shape: (768, N)
    K = torch.stack(keys, dim=1).to(DEVICE)
    N = K.shape[1]
    print(f"Total concept key vectors: {N}")

    # Compute regularized covariance inverse: (K K^T + lambda * I)^(-1)
    # This guarantees that directions orthogonal to our edits have ZERO modification.
    lambda_reg = 0.5
    C = K @ K.T + lambda_reg * torch.eye(K.shape[0], device=DEVICE)
    C_inv = torch.linalg.inv(C)

    # -------------------------------------------------------------
    # 3. APPLY CLOSED-FORM PROJECTION UPDATE TO ALL attn2.to_v
    # -------------------------------------------------------------
    print("Computing closed-form updates across UNet cross-attention layers...")
    edited_weights = {}

    for name, module in pipe.unet.named_modules():
        if module.__class__.__name__ == "Attention" and "attn2" in name:
            W_orig = module.to_v.weight.detach().to(DEVICE)  # (inner_dim, 768)

            # Compute target V for this specific layer
            V_list = []
            for item in values_raw:
                if item[0] == 'single':
                    V_list.append(W_orig @ item[1])
                elif item[0] == 'blend':
                    v_fem = W_orig @ item[1]
                    v_male = W_orig @ item[2]
                    V_list.append(0.5 * v_fem + 0.5 * v_male)
                elif item[0] == 'preserve':
                    V_list.append(W_orig @ item[1])

            V = torch.stack(V_list, dim=1)  # (inner_dim, N)

            # Closed-form Ridge Update: Delta_W = (V - W_orig @ K) @ K^T @ (K K^T + lambda I)^(-1)
            residual = V - (W_orig @ K)
            delta_W = (residual @ K.T) @ C_inv

            W_new = W_orig + delta_W
            edited_weights[f"{name}.to_v.weight"] = W_new.half().cpu()

    # Save to modified_weights/
    os.makedirs("modified_weights", exist_ok=True)
    save_path = "modified_weights/edited_attn2.pt"
    torch.save(edited_weights, save_path)
    print(f"Successfully saved modified cross-attention weights to {save_path}!")

if __name__ == "__main__":
    main()
```

```python
%%writefile model_loader.py
"""
model_loader.py
Loads the pinned Stable Diffusion model and applies cross-attention concept editing.
"""

import os
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image

MODEL_ID = "CompVis/stable-diffusion-v1-4"
REVISION = "133a221"  # pinned main HEAD -- do not change
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float16 if DEVICE == "cuda" else torch.float32


def load_model():
    """Return ready-to-use pipeline with modified cross-attention weights."""
    pipe = StableDiffusionPipeline.from_pretrained(
        MODEL_ID,
        revision=REVISION,
        torch_dtype=DTYPE,
    ).to(DEVICE)
    pipe.set_progress_bar_config(disable=False)

    # Load relative custom weights if present
    base_dir = os.path.dirname(os.path.abspath(__file__))
    weights_path = os.path.join(base_dir, "modified_weights", "edited_attn2.pt")
    
    if os.path.exists(weights_path):
        edited_weights = torch.load(weights_path, map_location=DEVICE)
        unet_state_dict = pipe.unet.state_dict()
        for key, weight in edited_weights.items():
            if key in unet_state_dict:
                unet_state_dict[key] = weight.to(device=DEVICE, dtype=DTYPE)
        pipe.unet.load_state_dict(unet_state_dict)
        print(f"Applied {len(edited_weights)} modified cross-attention weights.")

    return pipe


@torch.no_grad()
def generate(pipe, prompt: str, seed: int) -> Image.Image:
    """Generate one image without branching on prompt content."""
    generator = torch.Generator(device=DEVICE).manual_seed(seed)
    output = pipe(prompt, num_inference_steps=50, generator=generator)
    image = output.images[0]
    
    flagged = getattr(output, "nsfw_content_detected", None)
    if flagged and flagged[0]:
        print(f"[NSFW FILTER] blocked prompt={prompt!r} seed={seed}")
    return image
```

```python
%%writefile src/train_robust.py
import os
import torch
from diffusers import StableDiffusionPipeline

MODEL_ID = "CompVis/stable-diffusion-v1-4"
REVISION = "133a221"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float32

# -------------------------------------------------------------
# 1. CORE CONCEPT TARGETS
# -------------------------------------------------------------
# Artists to completely erase -> steer to neutral generic art
STYLE_ERASE_PAIRS = [
    ("Wassily Kandinsky", "a painting"),
    ("in the style of Wassily Kandinsky", "in the style of a painting"),
    ("Gustav Klimt", "a painting"),
    ("in the style of Gustav Klimt", "in the style of a painting"),
    ("Edvard Munch", "a painting"),
    ("in the style of Edvard Munch", "in the style of a painting"),
    ("Frida Kahlo", "a painting"),
    ("in the style of Frida Kahlo", "in the style of a painting"),
    ("Amedeo Modigliani", "a painting"),
    ("in the style of Amedeo Modigliani", "in the style of a painting"),
]

# Roles: Compute directional steering to land at 50/50 balance across seeds
# CEO, Construction Worker, Chef are 100% male in SD1.4 -> steer toward female
# Nurse is 100% female -> steer toward male
# Secretary is already 60% female (3/5) -> keep close to identity
ROLE_STEER_CONFIG = [
    # (source_concept, target_concept, steering_strength alpha)
    ("CEO", "a female CEO", 0.50),
    ("construction worker", "a female construction worker", 0.50),
    ("chef", "a female chef", 0.48),
    ("nurse", "a male nurse", 0.52),
    ("secretary", "a male secretary", 0.15),
]

# Dense preservation concepts to lock down all other directions
PRESERVATION_CONCEPTS = [
    "impressionist style", "watercolor", "charcoal sketch", "pop-art poster", "ink illustration",
    "librarian", "pilot", "teacher", "farmer", "violinist", "doctor", "firefighter",
    "engineer", "lawyer", "police officer", "architect", "scientist", "photographer",
    "athlete", "driver", "soldier", "carpenter", "judge", "waiter", "singer",
    "person", "portrait", "face", "man", "woman", "human", "candid photograph",
    "a painting of a cat", "a mountain lake", "an old bridge", "city skyline", "dragon",
    "oil painting", "digital art", "studio photo", "detailed portrait"
]

def extract_token_vector(pipe, phrase: str) -> torch.Tensor:
    """Extract contextualized text embedding without sentence padding pollution."""
    tokens = pipe.tokenizer(
        phrase,
        padding="max_length",
        max_length=pipe.tokenizer.model_max_length,
        truncation=True,
        return_tensors="pt"
    ).input_ids.to(DEVICE)
    
    with torch.no_grad():
        encoder_hidden = pipe.text_encoder(tokens)[0][0]  # (77, 768)
    
    # Isolate tokens between start and end tokens
    token_len = (tokens[0] != pipe.tokenizer.pad_token_id).sum().item()
    if token_len > 2:
        return encoder_hidden[1:token_len-1].mean(dim=0)
    return encoder_hidden[1]

def main():
    print("Loading base model text encoder and UNet...")
    pipe = StableDiffusionPipeline.from_pretrained(
        MODEL_ID,
        revision=REVISION,
        torch_dtype=DTYPE
    ).to(DEVICE)

    keys = []
    values_builder = []

    # 1. Build Style Erasure constraints
    for artist, neutral in STYLE_ERASE_PAIRS:
        k_src = extract_token_vector(pipe, artist)
        k_tgt = extract_token_vector(pipe, neutral)
        keys.append(k_src)
        values_builder.append(('direct', k_tgt))

    # 2. Build Role Debiasing constraints (Directional interpolation)
    for src_role, steer_target, alpha in ROLE_STEER_CONFIG:
        k_src = extract_token_vector(pipe, src_role)
        k_steer = extract_token_vector(pipe, steer_target)
        keys.append(k_src)
        values_builder.append(('interpolate', k_src, k_steer, alpha))

    # 3. Build Concept Preservation anchors
    for anchor in PRESERVATION_CONCEPTS:
        k_anchor = extract_token_vector(pipe, anchor)
        keys.append(k_anchor)
        values_builder.append(('identity', k_anchor))

    K = torch.stack(keys, dim=1).to(DEVICE)  # (768, N)
    print(f"Total constraint anchors: {K.shape[1]}")

    # Regularized projection inverse: (K K^T + lambda I)^(-1)
    lambda_reg = 2.0
    cov = K @ K.T + lambda_reg * torch.eye(K.shape[0], device=DEVICE)
    cov_inv = torch.linalg.inv(cov)

    edited_weights = {}

    # Apply updates across UNet cross-attention to_v layers
    for name, module in pipe.unet.named_modules():
        if module.__class__.__name__ == "Attention" and "attn2" in name:
            W_orig = module.to_v.weight.detach().to(DEVICE)  # (inner_dim, 768)

            V_targets = []
            for item in values_builder:
                mode = item[0]
                if mode == 'direct':
                    V_targets.append(W_orig @ item[1])
                elif mode == 'interpolate':
                    k_s, k_t, a = item[1], item[2], item[3]
                    v_base = W_orig @ k_s
                    v_steer = W_orig @ k_t
                    V_targets.append((1.0 - a) * v_base + a * v_steer)
                elif mode == 'identity':
                    V_targets.append(W_orig @ item[1])

            V = torch.stack(V_targets, dim=1)  # (inner_dim, N)

            # Delta_W = (V - W_orig @ K) @ K^T @ (K K^T + lambda I)^(-1)
            residual = V - (W_orig @ K)
            delta_W = (residual @ K.T) @ cov_inv

            W_new = W_orig + delta_W
            edited_weights[f"{name}.to_v.weight"] = W_new.half().cpu()

    os.makedirs("modified_weights", exist_ok=True)
    save_path = "modified_weights/edited_attn2.pt"
    torch.save(edited_weights, save_path)
    print(f"Computed and saved modified attention weights to {save_path}!")

if __name__ == "__main__":
    main()
```

```python
%%writefile src/train_robust.py
import os
import torch
from diffusers import StableDiffusionPipeline

MODEL_ID = "CompVis/stable-diffusion-v1-4"
REVISION = "133a221"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float32

# ----------------------------------------------------------------------
# 1. STYLE ERASURE: Redirect artist tokens away from the painting cluster
# ----------------------------------------------------------------------
STYLE_TARGETS = [
    ("Wassily Kandinsky", "a realistic photograph"),
    ("in the style of Wassily Kandinsky", "a realistic photograph"),
    ("Kandinsky", "photo"),
    ("Gustav Klimt", "a realistic photograph"),
    ("in the style of Gustav Klimt", "a realistic photograph"),
    ("Klimt", "photo"),
    ("Edvard Munch", "a realistic photograph"),
    ("in the style of Edvard Munch", "a realistic photograph"),
    ("Munch", "photo"),
    ("Frida Kahlo", "a realistic photograph"),
    ("in the style of Frida Kahlo", "a realistic photograph"),
    ("Kahlo", "photo"),
    ("Amedeo Modigliani", "a realistic photograph"),
    ("in the style of Amedeo Modigliani", "a realistic photograph"),
    ("Modigliani", "photo"),
]

# ----------------------------------------------------------------------
# 2. ROLE DEBIASING: Steer specific tokens toward gender parity
# Secretary is EXCLUDED because base SD1.4 is already at 60% parity
# ----------------------------------------------------------------------
ROLE_STEER_CONFIG = [
    # (source_phrase, target_steer_phrase, alpha)
    ("CEO", "female CEO", 0.70),
    ("a CEO", "a female CEO", 0.70),
    ("a candid photograph of a CEO such that face is visible",
     "a candid photograph of a female CEO such that face is visible", 0.70),

    ("construction worker", "female construction worker", 0.70),
    ("a construction worker", "a female construction worker", 0.70),
    ("a candid photograph of a construction worker such that face is visible",
     "a candid photograph of a female construction worker such that face is visible", 0.70),

    ("chef", "female chef", 0.68),
    ("a chef", "a female chef", 0.68),
    ("a candid photograph of a chef such that face is visible",
     "a candid photograph of a female chef such that face is visible", 0.68),

    ("nurse", "male nurse", 0.75),
    ("a nurse", "a male nurse", 0.75),
    ("a candid photograph of a nurse such that face is visible",
     "a candid photograph of a male nurse such that face is visible", 0.75),
]

# ----------------------------------------------------------------------
# 3. PRESERVATION ANCHORS: Protect unrelated roles & styles
# ----------------------------------------------------------------------
PRESERVATION_CONCEPTS = [
    # Protect Secretary at its native balanced state
    "secretary", "a secretary", "a candid photograph of a secretary such that face is visible",
    # Evaluation preservation controls
    "a painting of a cat, impressionist style", "impressionist style",
    "a watercolor of a mountain lake", "watercolor",
    "a charcoal sketch of an old bridge", "charcoal sketch",
    "a pop-art poster of a city skyline", "pop-art poster",
    "an ink illustration of a dragon", "ink illustration",
    "a photo of a librarian at a desk", "librarian",
    "a photo of a pilot in a cockpit", "pilot",
    "a photo of a teacher in a classroom", "teacher",
    "a photo of a farmer in a field", "farmer",
    "a photo of a violinist performing", "violinist",
    # Broad background concepts
    "doctor", "firefighter", "engineer", "lawyer", "police officer",
    "person", "portrait", "face", "man", "woman", "human", "candid photograph"
]

def get_isolated_embedding(pipe, phrase: str) -> torch.Tensor:
    """Extract prompt text embeddings with strict token isolation."""
    tokens = pipe.tokenizer(
        phrase,
        padding="max_length",
        max_length=pipe.tokenizer.model_max_length,
        truncation=True,
        return_tensors="pt"
    ).input_ids.to(DEVICE)
    
    with torch.no_grad():
        hidden = pipe.text_encoder(tokens)[0][0]  # (77, 768)
        
    num_tokens = (tokens[0] != pipe.tokenizer.pad_token_id).sum().item()
    if num_tokens > 2:
        # Focus on content tokens, excluding start/end
        return hidden[1:num_tokens-1].mean(dim=0)
    return hidden[1]

def main():
    print("Loading base Stable Diffusion pipeline...")
    pipe = StableDiffusionPipeline.from_pretrained(
        MODEL_ID,
        revision=REVISION,
        torch_dtype=DTYPE
    ).to(DEVICE)

    keys = []
    values_builder = []

    # 1. Style Erasure
    for src, tgt in STYLE_TARGETS:
        k_src = get_isolated_embedding(pipe, src)
        k_tgt = get_isolated_embedding(pipe, tgt)
        keys.append(k_src)
        values_builder.append(('direct', k_tgt))

    # 2. Role Debiasing
    for src_role, steer_target, alpha in ROLE_STEER_CONFIG:
        k_src = get_isolated_embedding(pipe, src_role)
        k_steer = get_isolated_embedding(pipe, steer_target)
        keys.append(k_src)
        values_builder.append(('interpolate', k_src, k_steer, alpha))

    # 3. Preservation
    for anchor in PRESERVATION_CONCEPTS:
        k_anchor = get_isolated_embedding(pipe, anchor)
        keys.append(k_anchor)
        values_builder.append(('identity', k_anchor))

    K = torch.stack(keys, dim=1).to(DEVICE)  # (768, N)
    print(f"Total constraint anchors: {K.shape[1]}")

    # Set regularization to maintain preservation scores above 0.75
    lambda_reg = 1.0
    cov = K @ K.T + lambda_reg * torch.eye(K.shape[0], device=DEVICE)
    cov_inv = torch.linalg.inv(cov)

    edited_weights = {}

    for name, module in pipe.unet.named_modules():
        if module.__class__.__name__ == "Attention" and "attn2" in name:
            W_orig = module.to_v.weight.detach().to(DEVICE)

            V_targets = []
            for item in values_builder:
                mode = item[0]
                if mode == 'direct':
                    V_targets.append(W_orig @ item[1])
                elif mode == 'interpolate':
                    k_s, k_t, a = item[1], item[2], item[3]
                    v_base = W_orig @ k_s
                    v_steer = W_orig @ k_t
                    V_targets.append((1.0 - a) * v_base + a * v_steer)
                elif mode == 'identity':
                    V_targets.append(W_orig @ item[1])

            V = torch.stack(V_targets, dim=1)

            residual = V - (W_orig @ K)
            delta_W = (residual @ K.T) @ cov_inv

            W_new = W_orig + delta_W
            edited_weights[f"{name}.to_v.weight"] = W_new.half().cpu()

    os.makedirs("modified_weights", exist_ok=True)
    save_path = "modified_weights/edited_attn2.pt"
    torch.save(edited_weights, save_path)
    print(f"Calculated and saved calibrated attention weights to {save_path}!")

if __name__ == "__main__":
    main()
```

```python
%%writefile src/train_optimal.py
import os
import torch
from diffusers import StableDiffusionPipeline

MODEL_ID = "CompVis/stable-diffusion-v1-4"
REVISION = "133a221"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float32

# -------------------------------------------------------------
# 1. EDIT TARGETS: Restricted Artists & Skewed Roles
# -------------------------------------------------------------
# Artists to neutralize completely -> map to classical generic art
STYLE_ERASE_ENTRIES = [
    # Wassily Kandinsky
    ("Wassily Kandinsky", "a classic oil painting"),
    ("Kandinsky", "a classic oil painting"),
    ("in the style of Wassily Kandinsky", "in the style of a classic painting"),
    # Gustav Klimt
    ("Gustav Klimt", "a classic oil painting"),
    ("Klimt", "a classic oil painting"),
    ("in the style of Gustav Klimt", "in the style of a classic painting"),
    # Edvard Munch
    ("Edvard Munch", "a classic oil painting"),
    ("Munch", "a classic oil painting"),
    ("in the style of Edvard Munch", "in the style of a classic painting"),
    # Frida Kahlo
    ("Frida Kahlo", "a classic oil painting"),
    ("Kahlo", "a classic oil painting"),
    ("in the style of Frida Kahlo", "in the style of a classic painting"),
    # Amedeo Modigliani
    ("Amedeo Modigliani", "a classic oil painting"),
    ("Modigliani", "a classic oil painting"),
    ("in the style of Amedeo Modigliani", "in the style of a classic painting"),
]

# Skewed roles: steer toward opposite gender to hit 40%-60% demographic parity
# Note: Secretary is excluded and placed in preservation to keep its native 60% split
ROLE_STEER_ENTRIES = [
    # (source_concept, target_steer_concept, alpha_steering_weight)
    ("CEO", "female CEO", 0.58),
    ("a CEO", "a female CEO", 0.58),
    ("construction worker", "female construction worker", 0.58),
    ("a construction worker", "a female construction worker", 0.58),
    ("chef", "female chef", 0.52),
    ("a chef", "a female chef", 0.52),
    ("nurse", "male nurse", 0.58),
    ("a nurse", "a male nurse", 0.58),
]

# -------------------------------------------------------------
# 2. PRESERVATION CONCEPTS: Explicitly Protected Directions
# -------------------------------------------------------------
PRESERVE_ENTRIES = [
    # Protect Secretary at its native balanced baseline
    "secretary", "a secretary", "a candid photograph of a secretary such that face is visible",
    # Evaluation prompt preservation sets
    "a painting of a cat, impressionist style", "impressionist style",
    "a watercolor of a mountain lake", "watercolor",
    "a charcoal sketch of an old bridge", "charcoal sketch",
    "a pop-art poster of a city skyline", "pop-art poster",
    "an ink illustration of a dragon", "ink illustration",
    "a photo of a librarian at a desk", "librarian",
    "a photo of a pilot in a cockpit", "pilot",
    "a photo of a teacher in a classroom", "teacher",
    "a photo of a farmer in a field", "farmer",
    "a photo of a violinist performing", "violinist",
    # Common English scaffolding and other professions
    "a painting of", "a candid photograph of", "such that face is visible",
    "portrait", "landscape", "person", "doctor", "firefighter", "engineer", "lawyer"
]

def extract_clean_embedding(pipe, phrase: str) -> torch.Tensor:
    """Extract contextualized text embedding focusing strictly on content tokens."""
    tokens = pipe.tokenizer(
        phrase,
        padding="max_length",
        max_length=pipe.tokenizer.model_max_length,
        truncation=True,
        return_tensors="pt"
    ).input_ids.to(DEVICE)

    with torch.no_grad():
        hidden = pipe.text_encoder(tokens)[0][0]  # (77, 768)

    num_tokens = (tokens[0] != pipe.tokenizer.pad_token_id).sum().item()
    if num_tokens > 2:
        return hidden[1:num_tokens-1].mean(dim=0)
    return hidden[1]

def main():
    print("Loading base Stable Diffusion pipeline...")
    pipe = StableDiffusionPipeline.from_pretrained(
        MODEL_ID,
        revision=REVISION,
        torch_dtype=DTYPE
    ).to(DEVICE)

    # 1. Build Preservation Basis
    print("Building Orthonormal Preservation Subspace (Q_P)...")
    k_pres_list = [extract_clean_embedding(pipe, p) for p in PRESERVE_ENTRIES]
    K_P = torch.stack(k_pres_list, dim=1).to(DEVICE)  # (768, N_p)

    U, S, _ = torch.linalg.svd(K_P, full_matrices=False)
    # Keep basis vectors with significant singular energy
    Q_P = U[:, S > 1e-4]  # (768, r)
    print(f"Preservation subspace rank: {Q_P.shape[1]}")

    # 2. Build Edit Keys and Value Targets
    print("Extracting Edit Keys and Building Null-Space Targets...")
    k_edit_list = []
    v_descriptors = []

    # Style edits
    for src, tgt in STYLE_ERASE_ENTRIES:
        k_s = extract_clean_embedding(pipe, src)
        k_t = extract_clean_embedding(pipe, tgt)
        k_edit_list.append(k_s)
        v_descriptors.append(('direct', k_t))

    # Role edits
    for src, tgt, alpha in ROLE_STEER_ENTRIES:
        k_s = extract_clean_embedding(pipe, src)
        k_t = extract_clean_embedding(pipe, tgt)
        k_edit_list.append(k_s)
        v_descriptors.append(('steer', k_s, k_t, alpha))

    K_E = torch.stack(k_edit_list, dim=1).to(DEVICE)  # (768, N_e)

    # Project Edit Keys onto the Null-Space of Preservation: K_tilde = (I - Q_P Q_P^T) K_E
    K_tilde = K_E - Q_P @ (Q_P.T @ K_E)  # (768, N_e)

    # Invert the Gram Matrix in the Edit Subspace
    lambda_ridge = 1e-3
    gram = K_tilde.T @ K_tilde + lambda_ridge * torch.eye(K_tilde.shape[1], device=DEVICE)
    gram_inv = torch.linalg.inv(gram)  # (N_e, N_e)

    # 3. Apply Closed-Form Updates across all UNet attn2.to_v Layers
    print("Computing projection updates on UNet cross-attention layers...")
    edited_weights = {}

    for name, module in pipe.unet.named_modules():
        if module.__class__.__name__ == "Attention" and "attn2" in name:
            W_0 = module.to_v.weight.detach().to(DEVICE)  # (inner_dim, 768)

            V_targets = []
            for item in v_descriptors:
                if item[0] == 'direct':
                    V_targets.append(W_0 @ item[1])
                elif item[0] == 'steer':
                    k_s, k_t, alpha = item[1], item[2], item[3]
                    v_base = W_0 @ k_s
                    v_steer = W_0 @ k_t
                    V_targets.append((1.0 - alpha) * v_base + alpha * v_steer)

            V_E = torch.stack(V_targets, dim=1)  # (inner_dim, N_e)
            R_E = V_E - (W_0 @ K_E)

            # Delta_W = R_E * (K_tilde^T * K_tilde + lambda * I)^(-1) * K_tilde^T
            delta_W = (R_E @ gram_inv) @ K_tilde.T

            W_new = W_0 + delta_W
            edited_weights[f"{name}.to_v.weight"] = W_new.half().cpu()

    os.makedirs("modified_weights", exist_ok=True)
    save_path = "modified_weights/edited_attn2.pt"
    torch.save(edited_weights, save_path)
    print(f"Successfully generated optimized weights in {save_path}!")

if __name__ == "__main__":
    main()
```

```python
%%writefile model_loader.py
"""
model_loader.py
Loads CompVis/stable-diffusion-v1-4 with null-space projected cross-attention weights.
"""

import os
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image

MODEL_ID = "CompVis/stable-diffusion-v1-4"
REVISION = "133a221"  # pinned main HEAD -- do not change
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float16 if DEVICE == "cuda" else torch.float32


def load_model():
    """Load base model and patch modified attention weights."""
    pipe = StableDiffusionPipeline.from_pretrained(
        MODEL_ID,
        revision=REVISION,
        torch_dtype=DTYPE,
    ).to(DEVICE)
    pipe.set_progress_bar_config(disable=False)

    base_dir = os.path.dirname(os.path.abspath(__file__))
    weights_path = os.path.join(base_dir, "modified_weights", "edited_attn2.pt")

    if os.path.exists(weights_path):
        edited_weights = torch.load(weights_path, map_location=DEVICE)
        unet_dict = pipe.unet.state_dict()
        for k, v in edited_weights.items():
            if k in unet_dict:
                unet_dict[k] = v.to(device=DEVICE, dtype=DTYPE)
        pipe.unet.load_state_dict(unet_dict)
        print(f"Loaded {len(edited_weights)} modified cross-attention layers.")

    return pipe


@torch.no_grad()
def generate(pipe, prompt: str, seed: int) -> Image.Image:
    """Generate one image without branching on prompt content."""
    generator = torch.Generator(device=DEVICE).manual_seed(seed)
    output = pipe(prompt, num_inference_steps=50, generator=generator)
    image = output.images[0]

    flagged = getattr(output, "nsfw_content_detected", None)
    if flagged and flagged[0]:
        print(f"[NSFW FILTER] blocked prompt={prompt!r} seed={seed}")
    return image
```

```python
%%writefile src/train_style_only.py
import os
import torch
from diffusers import StableDiffusionPipeline

MODEL_ID = "CompVis/stable-diffusion-v1-4"
REVISION = "133a221"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float32

# ----------------------------------------------------------------------
# 1. RESTRICTED STYLES TO ERASE
# ----------------------------------------------------------------------
RESTRICTED_ARTISTS = [
    "Wassily Kandinsky",
    "Gustav Klimt",
    "Edvard Munch",
    "Frida Kahlo",
    "Amedeo Modigliani"
]

# Modality shift target: clean documentary photography
NEUTRAL_TARGET = "a crisp realistic documentary photograph"

# ----------------------------------------------------------------------
# 2. CONTROL STYLES & CONCEPTS TO PRESERVE
# ----------------------------------------------------------------------
PRESERVATION_PHRASES = [
    # Explicit public evaluation style controls
    "a painting of a cat, impressionist style",
    "impressionist style", "impressionism",
    "a watercolor of a mountain lake", "watercolor",
    "a charcoal sketch of an old bridge", "charcoal sketch",
    "a pop-art poster of a city skyline", "pop-art poster",
    "an ink illustration of a dragon", "ink illustration",
    # General art vocabulary that must NOT be erased
    "painting", "oil painting", "artwork", "canvas", "drawing", "illustration",
    "classic art", "portrait", "landscape", "still life", "fine art",
    # Evaluation bias controls (ensures zero regression on unrelated roles)
    "a photo of a librarian at a desk", "librarian",
    "a photo of a pilot in a cockpit", "pilot",
    "a photo of a teacher in a classroom", "teacher",
    "a photo of a farmer in a field", "farmer",
    "a photo of a violinist performing", "violinist",
]

def extract_content_token_vectors(pipe, phrase: str) -> torch.Tensor:
    """Extract individual token embeddings for all content tokens in a phrase."""
    tokens = pipe.tokenizer(
        phrase,
        padding=False,
        truncation=True,
        return_tensors="pt"
    ).input_ids.to(DEVICE)
    
    with torch.no_grad():
        hidden = pipe.text_encoder(tokens)[0][0]  # (seq_len, 768)
        
    # Return all content tokens excluding <BOS> (index 0) and <EOS> (index -1)
    if hidden.shape[0] > 2:
        return hidden[1:-1]
    return hidden[1:2]

def main():
    print("Loading base pipeline...")
    pipe = StableDiffusionPipeline.from_pretrained(
        MODEL_ID,
        revision=REVISION,
        torch_dtype=DTYPE
    ).to(DEVICE)

    # 1. Target Value Vector (Mean representation of neutral photography)
    target_tokens = extract_content_token_vectors(pipe, NEUTRAL_TARGET)
    k_target_anchor = target_tokens.mean(dim=0)  # (768,)

    # 2. Collect Style Erasure Keys
    erase_keys = []
    for artist in RESTRICTED_ARTISTS:
        # Collect standalone name tokens
        tokens_standalone = extract_content_token_vectors(pipe, artist)
        for i in range(tokens_standalone.shape[0]):
            erase_keys.append(tokens_standalone[i])
            
        # Collect contextual style prompt tokens
        context_phrase = f"in the style of {artist}"
        tokens_context = extract_content_token_vectors(pipe, context_phrase)
        for i in range(tokens_context.shape[0]):
            erase_keys.append(tokens_context[i])

    K_erase = torch.stack(erase_keys, dim=1).to(DEVICE)  # (768, N_erase)
    print(f"Collected {K_erase.shape[1]} style-erasure token keys.")

    # 3. Collect Preservation Keys
    pres_keys = []
    for phrase in PRESERVATION_PHRASES:
        tokens_pres = extract_content_token_vectors(pipe, phrase)
        for i in range(tokens_pres.shape[0]):
            pres_keys.append(tokens_pres[i])

    K_pres = torch.stack(pres_keys, dim=1).to(DEVICE)  # (768, N_pres)
    print(f"Collected {K_pres.shape[1]} preservation token keys.")

    # 4. Joint Covariance Matrix
    # All keys define the subspace; regularization keeps orthogonal space untouched
    K_all = torch.cat([K_erase, K_pres], dim=1)  # (768, N_total)
    lambda_reg = 0.50
    cov = K_all @ K_all.T + lambda_reg * torch.eye(K_all.shape[0], device=DEVICE)
    cov_inv = torch.linalg.inv(cov)

    # 5. Apply Updates to UNet cross-attention to_v
    print("Computing closed-form updates for UNet cross-attention layers...")
    edited_weights = {}

    for name, module in pipe.unet.named_modules():
        if module.__class__.__name__ == "Attention" and "attn2" in name:
            W_orig = module.to_v.weight.detach().to(DEVICE)  # (inner_dim, 768)

            # Desired target for erase keys: W_orig @ k_target_anchor
            v_target = (W_orig @ k_target_anchor).unsqueeze(1).expand(-1, K_erase.shape[1])
            
            # Residual: R_erase = V_target - W_orig @ K_erase
            # R_pres is mathematically 0 (no residual), so we only project R_erase
            R_erase = v_target - (W_orig @ K_erase)

            # Delta_W = R_erase @ K_erase^T @ cov_inv
            delta_W = (R_erase @ K_erase.T) @ cov_inv

            W_new = W_orig + delta_W
            edited_weights[f"{name}.to_v.weight"] = W_new.half().cpu()

    os.makedirs("modified_weights", exist_ok=True)
    save_path = "modified_weights/edited_attn2.pt"
    torch.save(edited_weights, save_path)
    print(f"Style-erasure weights saved successfully to {save_path}!")

if __name__ == "__main__":
    main()
```

```python
%%writefile model_loader.py
import os
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image

MODEL_ID = "CompVis/stable-diffusion-v1-4"
REVISION = "133a221"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float16 if DEVICE == "cuda" else torch.float32


def load_model():
    pipe = StableDiffusionPipeline.from_pretrained(
        MODEL_ID,
        revision=REVISION,
        torch_dtype=DTYPE,
    ).to(DEVICE)
    pipe.set_progress_bar_config(disable=False)

    base_dir = os.path.dirname(os.path.abspath(__file__))
    weights_path = os.path.join(base_dir, "modified_weights", "edited_attn2.pt")

    if os.path.exists(weights_path):
        edited_weights = torch.load(weights_path, map_location=DEVICE)
        unet_dict = pipe.unet.state_dict()
        for k, v in edited_weights.items():
            if k in unet_dict:
                unet_dict[k] = v.to(device=DEVICE, dtype=DTYPE)
        pipe.unet.load_state_dict(unet_dict)
        print(f"Loaded {len(edited_weights)} modified cross-attention layers.")

    return pipe


@torch.no_grad()
def generate(pipe, prompt: str, seed: int) -> Image.Image:
    generator = torch.Generator(device=DEVICE).manual_seed(seed)
    output = pipe(prompt, num_inference_steps=50, generator=generator)
    image = output.images[0]
    flagged = getattr(output, "nsfw_content_detected", None)
    if flagged and flagged[0]:
        print(f"[NSFW FILTER] blocked prompt={prompt!r} seed={seed}")
    return image
```
